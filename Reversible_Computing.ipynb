**Reversible Computing**

***1. Reviersible XOR (X gate)***

```python
def x(a_in):
    a_out = a_in ^ 1  # ^1 perform the XOR operation 
    return a_out
```

```python
def cx(a_in, b_in):
    a_out = a_in
    
    # if control is one (i.e., a == 1): negate target (b' = b̅), 
    # else: leave b alone (b' = b).
    if a_in == 1:                
        b_out = x(b_in)
    else:
        b_out = b_in

    return (a_out, b_out)
```

```python
# Print table for CX gate
print('CX gate:')
print('a | b || a\' | b\' |')

for a, b in [(0,0), (0,1), (1,0), (1,1)]:
    ap, bp = cx(a,b)
    print(f'{a} | {b} || {ap}  | {bp}  |')
```

```python
# Print table for applying two CX gates
print('Two CX gates:')
print('a | b || a\'\' | b\'\' |')

# Apply CX twice for all combination of inputs a, b.
for a, b in [(0,0), (0,1), (1,0), (1,1)]:
    ai, bi = cx(a,b)
    ap, bp = cx(ai,bi)
    print(f'{a} | {b} || {ap}   | {bp}   |')
```
